## $k$ VS $\Delta T_{eff}$

mock stars: {$T_{eff} = \left\{3800, 7200\right\}$, $\log(g) = 4.5$, $Fe/H = 0.0$}

```python
d_Teff = [1600,800,400,200,100,50,25,10,5]
kTeff_Teff3800 = [4.97, 4.99, 4.95, 4.92, 4.80, 4.66, 4.51, 4.35, 4.23]
mTeff_Teff3800 = np.array([2.71, 2.41, 2.44, 2.42, 2.44, 2.11, 2.13, 2.11, 2.11])

kTeff_Teff7200 = [5, 5, 4.84, 4.67, 4.47, 4.34, 4.15, 4.06, 4.04]
mTeff_Teff7200 = np.array([3, 2.7, 2.44, 2.43, 2.12, 2.12, 1.82, 1.82, 1.82])


plt.figure()
plt.semilogx(d_Teff, kTeff_Teff3800, c='k', linestyle='-', label="k; Teff = 3800")
plt.semilogx(d_Teff, mTeff_Teff3800+2, c='k', linestyle='--', label="m; Teff = 3800")

plt.semilogx(d_Teff, kTeff_Teff7200, c='r', marker='o', linestyle='-', label="k; Teff = 7200")
plt.semilogx(d_Teff, mTeff_Teff7200+2, c='r', marker='o',linestyle='--', label="m; Teff = 7200")

plt.ylim(3.5, 5.2)
plt.xlabel(f"$\Delta Teff$")
plt.ylabel(f"$k$")
plt.legend()
plt.show()
```
